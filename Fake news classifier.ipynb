```python
from tensorflow.keras.layers import Embedding
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.preprocessing.text import one_hot
from tensorflow.keras.layers import LSTM
from tensorflow.keras.layers import Dense
from tensorflow.keras.layers import Bidirectional
```

```python
#vacobalury size
voc_size=5000
```

```python
from nltk.stem.porter import PorterStemmer
from nltk.corpus import stopwords
import re

ps = PorterStemmer()  # instantiate the PorterStemmer
corpus = []

for i in range(0 , len(messages)):
    print(i)
    review = str(messages['title'][i])  # convert to string to avoid the error
    review = re.sub('[^a-zA-Z]', ' ', review)
    review = review.lower()
    review = review.split()
    review = [ps.stem(word) for word in review if word not in stopwords.words('english')]
    review = ' '.join(review)
    corpus.append(review)
```

```python
corpus
```

```python
onehot_repr = [one_hot(words, voc_size) for words in corpus]
print(onehot_repr)
```

```python
#Embedding presentation
sent_length = 20
embedded_docs = pad_sequences(onehot_repr , padding = 'pre', maxlen = sent_length)
embedded_docs
```

```python
## Creating model
embedding_vector_features=40
model=Sequential()
model.add(Embedding(voc_size,embedding_vector_features,input_length=sent_length))
model.add(Bidirectional(LSTM(100)))
model.add(Dense(1,activation='sigmoid'))
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
print(model.summary())
```

```python
import numpy as np
x_final = np.array(embedded_docs)
y_final = np.array(y)
```

```python
##train test split 
from sklearn.model_selection import train_test_split
x_train, x_test , y_train, y_test = train_test_split(x_final, y_final, test_size = 0.2, random_state = 23)
```

```python
#model trainig 
model.fit(x_final, y_final, validation_data = (x_test, y_test), epochs = 10,batch_size = 32)
```

```python
y_pred = model.predict(x_test)
```

```python
import numpy as np

# If y_pred is a tuple, convert it to a numpy array
y_pred = np.array(y_pred)

# Now apply the threshold
y_pred = np.where(y_pred >= 0.5, 1, 0)

print(y_pred)
```
